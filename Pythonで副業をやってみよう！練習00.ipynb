内容は<br>
*   氏名
*   年齢
*   購入金額
*   購入日
*   カテゴリ


```python
import pandas as pd
import numpy as np

np.random.seed(0)
n = 100
df = pd.DataFrame({
    '氏名': [f'顧客{i}' for i in range(n)],
    '年齢': np.random.randint(20, 70, n),
    '購入金額': np.random.randint(1000, 50000, n),
    '購入日': pd.date_range('2025-01-01', periods=n, freq='D'),
    'カテゴリ': np.random.choice(['食品', '衣類', '家電', '書籍'], n)
})

df.to_excel('sample_data.xlsx', index=False)
```

今回は欠損値は無い<br>
では重複は？<br>
→重複確認は必要か？必要ないのでは？<br>
理由は単純に今回は加工が目的なのでだから。


データには2種類ある<br>
*   質的データ：場所や氏名や順番に意味のある数字なんかがこれ
*   量的データ：0に意味がある




```python
df['氏名'].head(10)
```

```python
df['年齢'].head(10)
```

```python
# ① 重複を除いた「種類の数」を確認(nunique = number of unique)
print(df['年齢'].nunique())
# 例: 3 → 全体の行数と比べて重複があるか判断できる

# ② 各値が何回登場するか確認(value_counts = 値ごとの件数)
print(df['年齢'].value_counts())
# 25    3
# 20    2
# 30    1

# ③ 重複している行そのものを見たい(duplicated = 重複判定。2回目以降がTrue)
print(df[df['年齢'].duplicated()])

# ④ 「重複が1件でもあるか」だけをTrue/Falseで知りたい場合
has_duplicates = df['年齢'].duplicated().any()
print(has_duplicates)
# True(1件でも重複があればTrue)

# ⑤ 全体の行数とユニーク数を比較する方法(直感的にわかりやすい)
print(f"全体: {len(df)}件, ユニーク: {df['年齢'].nunique()}件")
# 一致していれば重複なし、少なければ重複ありと判断できる
```

```python
# 重複している行を全部(1回目も含めて)表示したい場合
df[df['年齢'].duplicated(keep=False)]
```
